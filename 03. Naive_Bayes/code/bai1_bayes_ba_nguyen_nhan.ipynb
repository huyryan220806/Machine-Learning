Mục 7, bài 1: tính xác suất thuộc từng nhóm khi biết một bạn đã qua môn và kiểm tra tổng bằng 1.

Chọn **Run All** để chạy độc lập notebook này. Trên Colab, tải CSV cần dùng vào Files hoặc thư mục `data/`.


```python
from pathlib import Path

# Chạy từ thư mục bài, code/, hoặc thư mục chứa CSV đã tải lên Colab.
thu_muc_hien_tai = Path.cwd().resolve()
thu_muc_du_an = thu_muc_hien_tai
for thu_muc in (thu_muc_hien_tai, *thu_muc_hien_tai.parents):
    if (thu_muc / "data" / "tin_nhan.csv").is_file():
        thu_muc_du_an = thu_muc
        break

def tep_du_lieu(ten):
    for duong_dan in (thu_muc_du_an / "data" / ten, thu_muc_hien_tai / ten):
        if duong_dan.is_file():
            return duong_dan
    raise FileNotFoundError(f"Không tìm thấy {ten}. Hãy giữ thư mục data/ hoặc tải CSV lên Colab.")

thu_muc_hinh = thu_muc_du_an / "figures"
thu_muc_hinh.mkdir(exist_ok=True)
```

## Mã và kết quả

```python
nhom = ["Học đều", "Học dồn", "Ít học"]
tien_nghiem = [0.50, 0.30, 0.20]
ty_le_qua = [0.95, 0.70, 0.30]

def bayes_nhieu_nhom(prior, likelihood):
    tu_so = [p * l for p, l in zip(prior, likelihood)]
    bang_chung = sum(tu_so)
    return [gia_tri / bang_chung for gia_tri in tu_so]

hau_nghiem = bayes_nhieu_nhom(tien_nghiem, ty_le_qua)
p_qua = sum(p * l for p, l in zip(tien_nghiem, ty_le_qua))
print(f"P(qua môn) = 0.50×0.95 + 0.30×0.70 + 0.20×0.30 = {p_qua:.4f}")
for ten, p in zip(nhom, hau_nghiem):
    print(f"P({ten} | qua môn) = {p:.6f} ({p:.2%})")
print(f"Tổng xác suất hậu nghiệm = {sum(hau_nghiem):.6f}")
print("Tổng bằng 1: ba nhóm bao phủ toàn bộ lớp và không giao nhau.")
```
