# Bài tập 5 — Chọn sai biến thể Naive Bayes
So sánh MultinomialNB và GaussianNB trên hai cột giờ ôn và điểm giữa kỳ, giữ cách chia như mục 6.

Chọn **Run All** để chạy độc lập notebook này. Trên Colab, tải CSV cần dùng vào Files hoặc thư mục `data/`.


```python
from pathlib import Path

# Chạy từ thư mục bài, code/, hoặc thư mục chứa CSV đã tải lên Colab.
thu_muc_hien_tai = Path.cwd().resolve()
thu_muc_du_an = thu_muc_hien_tai
for thu_muc in (thu_muc_hien_tai, *thu_muc_hien_tai.parents):
    if (thu_muc / "data" / "tin_nhan.csv").is_file():
        thu_muc_du_an = thu_muc
        break

def tep_du_lieu(ten):
    for duong_dan in (thu_muc_du_an / "data" / ten, thu_muc_hien_tai / ten):
        if duong_dan.is_file():
            return duong_dan
    raise FileNotFoundError(f"Không tìm thấy {ten}. Hãy giữ thư mục data/ hoặc tải CSV lên Colab.")

thu_muc_hinh = thu_muc_du_an / "figures"
thu_muc_hinh.mkdir(exist_ok=True)
```

## Mã và kết quả

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.naive_bayes import GaussianNB

df = pd.read_csv(tep_du_lieu("sinh_vien.csv"))
X = df[["gio_on", "diem_giua_ky"]]
y = df["qua_mon"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

from sklearn.naive_bayes import MultinomialNB

for ten, model in [("MultinomialNB", MultinomialNB()), ("GaussianNB", GaussianNB())]:
    model.fit(X_train, y_train)
    acc = accuracy_score(y_test, model.predict(X_test))
    print(f"{ten:14s}: accuracy = {acc:.4f}")

print("Chương trình không báo lỗi vì hai cột đều không âm, đáp ứng kiểm tra đầu vào của MultinomialNB.")
print("Nhận xét: Giờ ôn và điểm giữa kỳ là dữ liệu liên tục, không phải số lần xuất hiện của các loại sự kiện. MultinomialNB coi chúng như số đếm nên dùng sai giả định thống kê dù vẫn chạy được.")
print("GaussianNB mô hình hóa trung bình và phương sai theo từng lớp, phù hợp với cách xử lý số liệu liên tục trong bài; accuracy trên một lần chia không thay thế việc kiểm tra giả định mô hình.")
```
